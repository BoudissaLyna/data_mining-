# Data Ingestion & Integration
**Objective:** Consolidate multiple data sources into a single raw integrated dataset.
**Project:** Laptop Price Prediction

---

```python
import pandas as pd
import numpy as np
import os

print("Libraries imported successfully")
```

```python
# Define paths to the source datasets
OUEDKNISS_PATH = 'dataset/ouedkniss_informatique_ordinateur_portable_20250811_180006.csv'
DATA_CSV_PATH = 'dataset/data.csv'
LAPTOP_DATA_PATH = 'dataset/laptop_price_prediction_cleaned.csv'

# 1.1 Load Ouedkniss
df_oued = pd.read_csv(OUEDKNISS_PATH)
oued_map = {
    'price_preview': 'PRICE', 'spec_Etat': 'LAPTOP_CONDITION', 
    'spec_Marque': 'LAPTOP_BRAND', 'spec_Processeur': 'CPU',
    'spec_RAM': 'RAM_SIZE', 'spec_Taille du disque': 'STORAGE_SIZE',
    'spec_Taille écran': 'SCREEN_SIZE', 'spec_Type disque': 'STORAGE_TYPE',
    'city': 'CITY'
}
df_oued = df_oued.rename(columns=oued_map)
df_oued['LAPTOP_MODEL'] = 'NeedToBeFilled'
df_oued['DEDICATED_GPU'] = 'NeedToBeFilled'
df_oued['GPU_GENERAL'] = 'NeedToBeFilled'
df_oued['GPU_INTEGRATED'] = 'NeedToBeFilled'
df_oued['SCREEN_FREQUENCY'] = 'NeedToBeFilled'
df_oued['SCREEN_RESOLUTION'] = 'NeedToBeFilled'
df_oued['RAM_TYPE'] = 'NeedToBeFilled'

# Extract year/month from ouedkniss created_at
df_oued['POST_YEAR'] = pd.to_datetime(df_oued['created_at']).dt.year
df_oued['POST_MONTH'] = pd.to_datetime(df_oued['created_at']).dt.month

# 1.2 Load Data.csv
df_data = pd.read_csv(DATA_CSV_PATH)
data_map = {'price_preview': 'PRICE', 'spec_Etat': 'LAPTOP_CONDITION', 'model_name': 'LAPTOP_MODEL', 'city': 'CITY'}
df_data = df_data.rename(columns=data_map)
df_data['LAPTOP_BRAND'] = 'NeedToBeFilled'
df_data['GPU_GENERAL'] = 'NeedToBeFilled'
df_data['GPU_INTEGRATED'] = 'NeedToBeFilled'
df_data['STORAGE_TYPE'] = 'NeedToBeFilled'
df_data['STORAGE_SIZE'] = 'NeedToBeFilled' # We have SSD/HDD columns here

# Extract year/month from data.csv created_at
df_data['POST_YEAR'] = pd.to_datetime(df_data['created_at']).dt.year
df_data['POST_MONTH'] = pd.to_datetime(df_data['created_at']).dt.month

# 1.3 Load Laptop Data
df_laptop = pd.read_csv(LAPTOP_DATA_PATH)
df_laptop['CITY'] = 'NeedToBeFilled'
df_laptop['GPU_GENERAL'] = 'NeedToBeFilled'
df_laptop['GPU_INTEGRATED'] = 'NeedToBeFilled'
df_laptop['STORAGE_TYPE'] = 'NeedToBeFilled'
df_laptop['STORAGE_SIZE'] = 'NeedToBeFilled'

# Final consolidation columns order
target_cols = [
    'PRICE', 'LAPTOP_CONDITION', 'LAPTOP_BRAND', 'LAPTOP_MODEL', 
    'DEDICATED_GPU', 'GPU_GENERAL', 'GPU_INTEGRATED', 'CPU',
    'RAM_SIZE', 'RAM_TYPE', 'SSD_SIZE', 'HDD_SIZE', 'STORAGE_SIZE', 'STORAGE_TYPE',
    'SCREEN_SIZE', 'SCREEN_FREQUENCY', 'SCREEN_RESOLUTION', 
    'CITY', 'POST_YEAR', 'POST_MONTH'
]

# Ensure all dataframes have these columns (fill missing with 'NeedToBeFilled')
for df_item in [df_oued, df_data, df_laptop]:
    for col in target_cols:
        if col not in df_item.columns:
            df_item[col] = 'NeedToBeFilled'

# Concatenate
df_integrated = pd.concat([df_oued[target_cols], df_data[target_cols], df_laptop[target_cols]], ignore_index=True)
print(f"Final Integrated Dataset: {df_integrated.shape[0]} rows")

# Save raw integrated dataset
df_integrated.to_csv('integrated_raw_dataset.csv', index=False)
print("Saved as: integrated_raw_dataset.csv")
```
