#  Team Data Collection & Integration Plan

**Objective**
Turn 66,667 raw Ouedkniss laptop listings into a single, reliable dataset, while making sure every team member’s work actually adds value instead of getting lost in a big merge.

---

##  Member Contributions & Roles

How We Split the Work: 

We followed a specialized cleaning approach.
Instead of everyone touching everything, each person became responsible for one part of the laptop — like a real assembly line.

### 1. Lyna: Core Market Identity
Handled: Price, brand, model, condition

Lyna focused on the heart of the market: price and identity.
She removed obviously fake prices (like 1 DZD or 999 million DZD) and cleaned brand names so that “DELL”, “Dell”, and “dell” didn’t end up as three different brands.

This step was crucial, because if the price logic is wrong, nothing else matters..

### 2. Aya: Visual & Display Metrics
Handled: Screen size, resolution, refresh rate

Aya worked on one of the messiest parts of the data: screens.
She converted text-based resolutions like “1920x1080” into usable numeric values, standardized refresh rates, and filtered out unrealistic screen sizes.

Anything outside the 11”–18” range was treated as noise.

### 3. Leena: Storage & Memory Architecture
Handled: RAM, SSD, HDD, storage type
Leena cleaned the core performance specs.
She fixed unit inconsistencies (GB vs TB), standardized storage types, and used market logic to fill strategic missing values.

For example, if a laptop was clearly a gaming model, it was very unlikely to be HDD-only — so SSD presence was inferred when reasonable.

### 4. Abdallah & Mimoun: Processing Power & Cross-Validation
Handled: CPU, dedicated GPU, integrated GPU

This duo acted as the reality check.
They validated whether CPU and GPU combinations actually made sense and removed impossible configurations (like an entry-level CPU paired with an RTX 4090).

They also helped align hardware specs with tier logic, reinforcing consistency across the dataset.


##  The Integration Workflow
Our integration process followed a clear, repeatable workflow:

Raw Ingestion
All scraped listings were combined into a single master file.

Specialized Cleaning Passes
Each team member applied their own cleaning logic only to their assigned features.

Cross-Feature Validation
We ran checks to ensure features made sense together — for example:

Does the price match the CPU tier?

Does the GPU match the laptop category?

Final Merge
All cleaned and validated features were merged into one Final Master Dataset.

##  Integration Results
- **Initial listings**: 66,667

Removed as noise or outliers: ~24,000

Final validated dataset: 42,419 high-quality records
