##  Objective
The goal of this phase is to transform raw cleaned data into "High-Signal" features that capture the complex relationships between hardware specifications and market value.

---

##  1. Physical Clarity (Screen Metrics)
- **Problem:** Resolution strings like "1920x1080" are categorical and do not capture "density".
- **Solution:** 
  - **Total Pixels:** `Width * Height` (measures sheer panel quality).
  - **PPI (Pixels Per Inch):** `sqrt(w^2 + h^2) / ScreenSize` (measures visual sharpness).

---

##  2. Speed over Space (Storage Metrics)
- **Problem:** 1TB HDD is cheaper than 256GB SSD, but raw "Size" says otherwise.
- **Solution:** 
  - **Storage Score:** `(SSD_GB * 1.0) + (HDD_GB * 0.2)`. This weighs SSD 5x more than HDD, reflecting its market impact.

---

##  3. High-Level Tiers (Logic Compression)
- **CPU_TIER:** Groups hundreds of CPU models into 4 distinct performance levels (*Entry, Mid, High, Enthusiast*).
- **BRAND_TIER:** Segments manufacturers into *Budget, Mainstream, and Premium* tiers based on median market pricing.

---

##  4. Market Identifiers
- **IS_GAMING:** A binary flag (1/0) that identifies high-performance machines based on model names (e.g., ROG, Legion, Alienware) and screen frequencies (>60Hz).

---

##  5. Output
Enriched dataset with **35 features** (up from 25 original columns), saved for model training and advanced analytics.
