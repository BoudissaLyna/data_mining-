## 1. Data Preparation & Splitting
We start by loading the enriched dataset from Phase 5 and splitting it into Training (70%) and Testing (30%) sets. This ensures consistent data for all team members.

```python
import pandas as pd
from sklearn.model_selection import train_test_split
import os

# Load the enriched dataset from the Results folder
input_path = 'D_RESULTS/final_cleaned_dataset.csv'

if os.path.exists(input_path):
    df = pd.read_csv(input_path)
    print(f"Loaded Enriched Dataset with {len(df)} records.")
    
    # Standard 70/30 split with fixed random state for reproducibility
    train_df, test_df = train_test_split(df, test_size=0.3, random_state=42)
    
    # Save the splits for individual model notebooks into the Results folder
    train_df.to_csv('D_RESULTS/training_dataset.csv', index=False)
    test_df.to_csv('D_RESULTS/testing_dataset.csv', index=False)
    
    print("SUCCESS: Data split and saved as 'training_dataset.csv' and 'testing_dataset.csv' in D_RESULTS folder.")
else:
    print(f"ERROR: {input_path} not found. Please run Phase 5 (Feature Engineering) first.")
```
