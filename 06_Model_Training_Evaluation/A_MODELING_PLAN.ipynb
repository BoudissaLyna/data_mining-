##  1. Project Objective
Our goal was to move beyond simple "price guessing" and create a comprehensive **Market Intelligence Suite** for the Algerian laptop market (Ouedkniss). We realized that price is not just a number, but a combination of brand identity, hardware logic, and seller trust.

##  2. The Four-Pillar Strategy
We applied four distinct Data Mining techniques to cover every angle of the market:

### A. Pattern Discovery (**Leena** - Association Rules)
*   **Method**: FP-Growth Algorithm.
*   **Logic**: Instead of looking at individual specs, we look at **"Sets"**. We wanted to find the "Laws of the Market" (e.g., *If Hardware A + Hardware B, then Price Tier C*).
*   **Goal**: To build a rulebook that explains why certain laptops are priced the way they are.

### B. Strategic Segmentation (**Aya** - Clustering)
*   **Method**: K-Means Clustering + PCA.
*   **Logic**: We let the data organize itself. We didn't tell the model what the categories were; we asked the model to find the natural "Groups" or "Tribes" of laptops in Algiers.
*   **Goal**: To identify market archetypes (e.g., The "Mass Market" vs. The "Elite Niche").

### C. Price Intelligence (**Lyna** - Classification)
*   **Method**: Decision Tree Classifier (Gini Index).
*   **Logic**: This is our **Primary Prediction Engine**. By using classification instead of regression, we created a model that is robust to the fluctuating Algerian Dinar, focusing on "Market Identity."
*   **Goal**: To accurately assign any laptop spec-set to one of four Market Tiers.

### D. Market Integrity (**Abdallah Mimoun** - Anomaly Detection)
*   **Method**: Isolation Forest & Local Outlier Factor (LOF).
*   **Logic**: Ouedkniss is a peer-to-peer market, which means there are scams and outliers. We used "Distance-based" and "Density-based" math to find listings that "don't belong."
*   **Goal**: To assign a **Trust Score** to every listing and filter out potential fraud.
