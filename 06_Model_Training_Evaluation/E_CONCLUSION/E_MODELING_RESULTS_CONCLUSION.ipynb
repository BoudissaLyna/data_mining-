## 🧬 2. The "Spec-Price" DNA (Rules & Clusters Sync)
*   **Leena’s** Association Rules proved that a laptop with **32GB RAM** has a high "Lift" (law-like connection) to "High-End" prices and SSD storage.
*   **Aya’s** Clustering independently confirmed this by putting these exact specs into a tiny, "Elite" group (Cluster 0) where the price doubles (~222k DZD).
*   **Decision Impact**: If a laptop is listed with 32GB RAM but has a "Budget" price, our **Anomaly Detection** will immediately flag it as a scam (Low Trust Score).

## 🛡️ 3. The Two Methods of Defense
We provide two distinct ways to look into the future:
1.  **Categorical Guidance (Classification)**: Tells you what "Market Identity" a laptop *should* have based on its specs. (e.g., "This spec belongs to the High-End tier").
2.  **Integrity Check (Anomaly Detection)**: Tells you if the listing is "too good to be true" or mathematically weird. It separates legitimate high-value items from statistical outliers or potential scams.
