#  Regression Part 1: Training (Lyna)
**Objective:** Train a DecisionTreeRegressor to predict the exact price (DZD).



---

```python
import pandas as pd
import joblib
from sklearn.tree import DecisionTreeRegressor
from sklearn.preprocessing import LabelEncoder

# 1. Load Data
train = pd.read_csv('../../D_RESULTS/training_dataset.csv')
test = pd.read_csv('../../D_RESULTS/testing_dataset.csv')

# 2. Features
features = ['LAPTOP_BRAND', 'LAPTOP_MODEL', 'CPU', 'RAM_GB', 'STORAGE_SCORE', 
            'IS_GAMING', 'PPI', 'CPU_TIER', 'BRAND_TIER', 'TOTAL_PIXELS']

X_train = train[features].copy()
y_train = train['PRICE']
X_test = test[features].copy() # needed for consistent encoding

# 3. Encoding
reg_encoders = {}
for col in X_train.select_dtypes(include=['object']).columns:
    le = LabelEncoder()
    full_cat = pd.concat([X_train[col], X_test[col]]).astype(str)
    le.fit(full_cat)
    X_train[col] = le.transform(X_train[col].astype(str))
    reg_encoders[col] = le

# 4. Training using MSE (Mean Squared Error / SSE logic)
reg_model = DecisionTreeRegressor(
    criterion='squared_error', 
    max_depth=35, 
    min_samples_leaf=2,
    random_state=42
)

reg_model.fit(X_train, y_train)

# 5. Save artifacts
joblib.dump(reg_model, '../../D_RESULTS/laptop_regression_model.pkl')
joblib.dump(reg_encoders, '../../D_RESULTS/regression_encoders.pkl')
joblib.dump(features, '../../D_RESULTS/regression_features.pkl')

print(" Regression Training Complete. Model saved.")
```
