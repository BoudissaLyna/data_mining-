##  Detailed Cleaning Methodology

### Phase 1: Individual Feature Cleaning
To ensure high data precision, each team member managed a specific technical domain. You can find the individual cleaning scripts and logic in the following paths:

-   **Lyna (Core Identity):** `03_Data_Preparation_Cleaning/C_TEAM_INDIVIDUAL_TASKS/Lyna_cleaning_task/`  
    *Focus: Price outliers, Brand standardization, and Condition normalization.*

-   **Aya (Display Metrics):** `03_Data_Preparation_Cleaning/C_TEAM_INDIVIDUAL_TASKS/Aya's_cleaning_task/`  
    *Focus: Screen size, Frequency extraction, and Resolution numeric conversion.*

-   **Leena (Storage Architecture):** `03_Data_Preparation_Cleaning/C_TEAM_INDIVIDUAL_TASKS/Leena's_cleaning_task/`  
    *Focus: RAM & Storage capacity extraction and Type classification.*

-   **Abdallah  Mimoun (Processing Power):** `03_Data_Preparation_Cleaning/C_TEAM_INDIVIDUAL_TASKS/Abdallah_Mimoun's_cleaning_task/`  
    *Focus: CPU family parsing, GPU tiers, and cross-hardware validation.*

### Phase 2: Cross-Feature Validation
- **Logic Checks:** Validated CPU-GPU compatibility (e.g., checking if price aligns with CPU tier).
- **Inference:** If CPU was missing but Model was known (e.g., Macbook Air M2), we filled the spec based on known defaults.

### Phase 3: Final Integration
- Merged cleaned columns into the master dataset using `03_Data_Preparation_Cleaning/B_MASTER_CLEANING.ipynb`.
- Re-verified price distribution to ensure clean target variables.

##  Feature Inventory

### Cleaned Features
1. `PRICE` - Target variable in DZD
2. `LAPTOP_CONDITION` - New/Used/Refurbished
3. `LAPTOP_BRAND` - Normalized manufacturer names
4. `LAPTOP_MODEL` - Standardized model strings
5. `CPU`, `DEDICATED_GPU`, `GPU_INTEGRATED` - Hardware specs
6. `RAM_SIZE`, `STORAGE_SIZE`, `SCREEN_RESOLUTION` - Raw text specs

### Engineered Numeric Features
These features were extracted to be used directly by our ML models:
- `RAM_GB`: Float capacity
- `SSD_GB`: Numeric SSD storage
- `HDD_GB`: Numeric HDD storage
- `SCREEN_FREQUENCY_NUM`: Continuous refresh rate value
