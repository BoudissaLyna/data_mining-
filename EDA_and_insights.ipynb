```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

df = pd.read_csv('final_cleaned_dataset.csv')
print(f"Dataset Rows: {df.shape[0]}")
print(f"Features Available: {df.shape[1]}")
```

## 1. Target Variable Analysis: PRICE
The target variable `PRICE` is the most critical. Let's look at its distribution.

```python
plt.figure(figsize=(10, 5))
sns.histplot(df['PRICE'], bins=50, kde=True, color='teal')
plt.title('Price Distribution in DZD')
plt.xlabel('Price (DZD)')
plt.ylabel('Frequency')
plt.show()

print(f"Median Price: {df['PRICE'].median():,.0f} DZD")
print(f"Mean Price: {df['PRICE'].mean():,.0f} DZD")
```

**Insight:** The price distribution is right-skewed. Most laptops are under 200,000 DZD, but there is a long tail of high-end gaming and professional machines reaching up to 800,000 DZD.

## 2. Feature Interactions: RAM & Performance
How does RAM and our new `STORAGE_SCORE` relate to price?

```python
plt.figure(figsize=(10, 6))
sns.boxplot(x='RAM_GB', y='PRICE', data=df, palette='viridis')
plt.title('Price vs RAM Capacity')
plt.show()
```

**Insight:** RAM is a strong non-linear predictor. The jump from 8GB to 16GB significantly increases the median price, and 32GB machines occupy the premium tier.

## 3. Brand & Market Tiers
Using our new `BRAND_TIER` and `CPU_TIER` features.

```python
fig, ax = plt.subplots(1, 2, figsize=(16, 6))

sns.barplot(x='BRAND_TIER', y='PRICE', data=df, ax=ax[0], order=['Premium', 'Mainstream', 'Budget', 'Niche'])
ax[0].set_title('Average Price by Brand Tier')

sns.barplot(x='CPU_TIER', y='PRICE', data=df, ax=ax[1], order=['Enthusiast', 'High-End', 'Mid-Range', 'Entry-Level'])
ax[1].set_title('Average Price by CPU Tier')

plt.show()
```

**Insight:** Our engineered tiers work! Enthusiast CPUs (i9/Ryzen 9) and Premium Brands (Apple/Razer) show a clear and significant price hierarchy.

## 4. The Gaming Impact
Is the `IS_GAMING` flag we created significant?

```python
plt.figure(figsize=(8, 6))
sns.violinplot(x='IS_GAMING', y='PRICE', data=df)
plt.title('Price Distribution: Office vs Gaming Laptops')
plt.xticks([0, 1], ['Office/General', 'Gaming/Pro'])
plt.show()
```

**Insight:** Gaming laptops have a much higher ceiling and variance. Most standard office laptops cluster tightly around the 60k-120k range.

## 5. Correlation Heatmap (Engineered Features)
Let's see how our new features correlate with Price.

```python
numeric_cols = ['PRICE', 'RAM_GB', 'SSD_GB', 'TOTAL_STORAGE_GB', 'STORAGE_SCORE', 
                'PPI', 'TOTAL_PIXELS', 'RAM_STORAGE_RATIO', 'IS_GAMING']
corr = df[numeric_cols].corr()

plt.figure(figsize=(10, 8))
sns.heatmap(corr, annot=True, cmap='coolwarm', fmt='.2f')
plt.title('Correlation Matrix of Key Features')
plt.show()
```

##  Final EDA Conclusions

1. **Price Drivers:** `RAM_GB`, `STORAGE_SCORE`, and `IS_GAMING` are the strongest numerical/flag correlations with price.
2. **Non-Linear Relationships:** Features like CPU and Brand are better handled as categorical Tiers (which we engineered).
3. **Resolution & PPI:** While PPI is a good metric, raw resolution (Total Pixels) has a stronger correlation with price as it indicates modern, premium screens.
4. **Outliers:** The dataset is now clean of \"impossible\" records, but we still have high-variance in the premium segment (Gaming/Pro), which will require robust modeling (like Random Forest or XGBoost).
