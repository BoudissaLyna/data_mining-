#  The Ouedkniss Laptop Journey: Our Quest for Price Logic

---

##  Phase 1: Problem Framing (Our Motivation)
When we first started this project, we asked ourselves: *"Why is it so hard to find a fair price for a laptop in Algeria?"*

The Ouedkniss marketplace is a digital jungle. Prices fluctuate wildly based on seller mood, urgent needs, and unstandardized specs. We decided to take on the challenge of **decoding this market**.

**Our Goal:** To build an AI model that doesn't just guess a number, but understands the *market tier* of a laptop based on its guts (RAM, CPU, Screen, and Storage). We chose **Classification** over simple Regression because we realized that in Algeria, laptops exist in "Market Identities" (e.g., "The Student Budget" vs "The Gamer's Choice").

---

##  Phase 2: Data Collection & The Great Integration
Our journey began with a massive pull of **66,667 raw listings**.

**What we did:**
1.  **Unified our Front:** We each took a part of the specs. Lyna handled the core identities (Brand/Price), Aya focused on the visual impact (Screens), Leena looked at the long-term memory (Storage), and Abdallah Mimoun validated the brainpower (CPU/GPU).
2.  **The "Unknown" Battle:** The raw data was full of placeholders like *"NeedToBeFilled"* or *"Prochainement"*. Instead of just deleting everything, we cross-referenced. If a laptop said 'i7-12700H', we *knew* what it was even if the seller didn't fill the CPU TIER column.
3.  **The Master Merge:** We integrated these separate findings into one 'Master Cleaned Dataset'. We ended up with **~42,000 pristine records**, having filtered out the "garbage" (laptops priced at 1 DZD or 999,999,999 DZD).

---

##  Phase 3: Feature Engineering (Speaking the AI's Language)
We didn't just throw raw numbers at the model. We engineered new "logic features":
-   **PPI & Total Pixels:** Because a clear screen is worth more.
-   **Storage Score:** We weighted SSD 5x more than HDD, reflecting real-world speed preferences.
-   **Gaming Flags:** We scanned model names for keywords like 'ROG' or 'Legion' to catch that premium "Gamer Tax" value.

---

##  Phase 4: Our Modeling Strategy
At this stage, we realized that one model was not enough.
Predicting a laptop’s tier is useful, but it does not explain why the market behaves the way it does, nor does it protect users from abnormal listings.

So instead of building a single predictor, we designed a four-model framework, where each model answers a different market question. Together, they form a complete view of the Ouedkniss laptop ecosystem.


## 1/ Classification — Defining the Market Identity

Question answered: “Which tier does this laptop belong to?”
This is our main predictive model.
We categorized laptops into Budget, Mainstream, High-End, and Ultimate, reflecting how sellers and buyers already think in practice.

Input: CPU, GPU, RAM, storage, screen, brand

Output: Market tier

This model is the backbone of the system


##  2/clustering _Discovering Natural Market Groups

Question answered: “What types of laptops exist, regardless of price?”
Clustering allowed the data to speak for itself, without predefined labels.
It revealed natural segments such as:
Gaming vs office laptops
Lightweight productivity vs performance-focused machines
This step helped us verify whether our tiers align with real usage patterns, and explained why laptops in the same tier may still feel very different.



##  3/Association Rules — Explaining the Hidden Pricing Logic
Question answered: “Which hardware combinations create value?”
Instead of predictions, this model produces human-readable rules, such as:
If GPU = RTX and RAM ≥ 16GB → likely High-End or Ultimate
If SSD + Full HD screen → minimum Mainstream tier
These rules turn the market logic into clear explanations, making the system interpretable for non-technical users.



##  Anomaly Detection — Protecting Trust in the Market
Question answered: “Does this listing make sense?”
This model identifies:
Overpriced laptops
Suspiciously cheap listings
Possible scams or incorrect specifications
It adds a trust layer to the system, ensuring that predictions are not blindly accepted when a listing is abnormal.

##  Unified Training & Evaluation Strategy
To ensure fairness and consistency across all models:
Train/Test Split: 70% / 30%
Shared preprocessing pipeline:
Encoding categorical features
Scaling numerical features (especially for clustering)
Class balancing when needed
Each team member focused on one modeling technique, but all models were trained and evaluated within the same framework, ensuring coherence and comparability.
