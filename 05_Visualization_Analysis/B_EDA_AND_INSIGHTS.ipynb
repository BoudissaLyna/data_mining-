# Exploratory Data Analysis (EDA) & Insights
**Objective:** Visualize relationships in the engineered dataset.
**Input:** `engineered_dataset.csv` (from Phase 04)

---

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import os

# Read the data resulting from Feature Engineering (Phase 04)
data_path = '../04_Feature_Engineering/engineered_dataset.csv'
df = pd.read_csv(data_path)
print(f"Dataset Rows: {df.shape[0]}")
```

```python
plt.figure(figsize=(10, 5))
sns.histplot(df['PRICE'], bins=50, kde=True, color='teal')
plt.title('Price Distribution in DZD')
plt.show()
```

```python
plt.figure(figsize=(10, 6))
sns.boxplot(x='BRAND_TIER', y='PRICE', data=df, palette='viridis')
plt.title('Price by Brand Tier')
plt.show()
```

```python
plt.figure(figsize=(8, 6))
sns.violinplot(x='IS_GAMING', y='PRICE', data=df)
plt.title('Price: Office vs Gaming')
plt.show()
```
