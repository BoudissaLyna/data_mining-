##  Part 1: Feature Glossary (35 Features)

This dataset contains **35 features** structured to provide a 360-degree view of the laptop market. We have categorized them into **Original Features** and **Engineered Features**.

### 🔹 Original Features (25)
These were cleaned and standardized by the team (**Lyna, Aya, Leena, Abdallah & Mimoun**).

| Feature | Type | Description |
| :--- | :--- | :--- |
| `PRICE` | Numeric | **Target Variable.** Price in Algerian Dinars (DZD). Range: 5k - 800k. |
| `LAPTOP_CONDITION` | Categorical | Condition: *New, Used, Refurbished*. |
| `LAPTOP_BRAND` | Categorical | Manufacturer (HP, Dell, Apple, etc.). |
| `LAPTOP_MODEL` | Categorical | Specific model series (e.g., Latitude, XPS, Pavilion). |
| `CPU` | Categorical | Processor string (e.g., "Intel Core i7 12th Gen"). |
| `RAM_GB` | Numeric | Memory capacity in GB. |
| `SSD_GB` | Numeric | SSD storage capacity. |
| `HDD_GB` | Numeric | HDD storage capacity. |
| `STORAGE_TYPE` | Categorical | Configuration: *SSD, HDD, or Hybrid*. |
| `SCREEN_SIZE` | Numeric | Diagonal size in inches. |
| `SCREEN_RESOLUTION` | String | Standardized resolution (e.g., "1920x1080"). |
| `SCREEN_FREQUENCY_NUM` | Numeric | Numeric refresh rate (60, 120, 144, etc.). |
| `CITY` | Categorical | Listing location (Algiers, Oran, etc.). |
| `POST_YEAR` | Numeric | Listing year. |
| `POST_MONTH` | Numeric | Listing month. |

###  Engineered Features (10)
These features were created to help our models understand complex market relationships.

| Feature | Goal | Logic |
| :--- | :--- | :--- |
| `RES_WIDTH` / `RES_HEIGHT` | Math | Numeric resolution components. |
| `TOTAL_PIXELS` | Quality | Higher pixels = Higher price floor. |
| `PPI` | Precision | **Pixels Per Inch.** Measures screen sharpness. |
| `TOTAL_STORAGE_GB` | Capacity | Sum of all storage types. |
| `IS_GAMING` | Split | Flag (1/0) identifying high-performance gaming hardware. |
| `CPU_TIER` | Simplicity | Groups CPUs into: *Entry, Mid, High, Enthusiast*. |
| `BRAND_TIER` | Simplicity | Groups brands by market position: *Budget, Mainstream, Premium*. |
| `STORAGE_SCORE` | Market | Values SSD 5x more than HDD to reflect real-world speed impact. |

##  Part 2: Key EDA Findings & Insights

After analyzing the visuals in our **B_EDA_NOTEBOOK**, we identified the following market trends:

### 1. Price Distribution & Market Stability
- Most listings cluster between **50,000 DZD and 150,000 DZD**. This is the "Mainstream" heart of the Algerian market.
- Listings above 250,000 DZD are highly sensitive to **CPU Tier** and **GPU** specs.

### 2. The Spec Premium
- **RAM Rule:** Moving from 8GB to 16GB RAM is the most consistent and valuable hardware upgrade.
- **SSD Dominance:** Even a small SSD drives a higher price premium than a large but slow HDD. Our `STORAGE_SCORE` proved this linear value increase.

### 3. Screen Sharpness (The PPI Factor)
- Higher resolution screens (4K/Retina) are a primary indicator of the **Ultimate Tier**. Our PPI vs Price analysis shows that screen clarity is a key driver for high-end pricing.

### 4. Correlation Highlights
- `RAM_GB`, `STORAGE_SCORE`, and `IS_GAMING` have the strongest positive correlations with Price.
- **Counter-Intuitive Find:** `SCREEN_SIZE` has a low correlation with price, proving that in Algeria, **specs and performance matter far more than screen real estate.**
